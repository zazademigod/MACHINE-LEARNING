```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor as dt
from sklearn.model_selection import train_test_split as tts
from sklearn.metrics import mean_absolute_error as mae
```

```python
sample = pd.DataFrame({"Names":["David", "Ndubuisi"], "Age":["23","24"]})
print(sample)
```

```python
names = ['tolu', 'ade', 'dele', 'fola', 'tunji', 'yemi', 'timi', 'ade', 'ayo', 'kunle', 'peter', 'banjo']
scores= [22,23,44,55,67,28,45,55,24,23,28,50]
ages= [12, 13, 12, 13, 12, 14, 15, 16, 12, 13 , 12, 15]
allowance= [75.03, 80.25, 90.7, 105.9, 150.5, 84.02, 90.4, 97.85, 70, 75.46, 76.84, 89.92]
```

```python

student_data = pd.DataFrame({'Name':names, 'Score':scores ,  'Age': ages, 'Allowance':allowance}, index = range(1, 13)) 
print(student_data)
```

```python
student_data.columns
student_data.to_csv('student data')
test= pd.read_csv('student data')
print(test.columns
     )
```

```python
y= test.Allowance
test_features = ['Age', 'Allowance']
X = test[test_features]
```

```python
train_X, val_X, train_y, val_y = tts(X, y, random_state=1)
s_model = DecisionTreeRegressor()
s_model.fit(train_X, train_y)
```

```python
pred= s_model.predict(val_X)
print(pred)
mae(val_y, pred)
```
